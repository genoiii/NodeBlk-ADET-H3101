```python
# Dummy test
#Register shipment

txn = contract.functions.registerShipment(
    "TEST-001", # rfidTag
    5, # GoodsCategory (DryGoods)
    "Manila", # origin
    "Cebu", # destination
    10, # packageCount
    "VH-TEST", # vehicleID
    "DR-TEST" # driverID
).transact({'from': w3.eth.default_account, 'gas': 1000000})
w3.eth.wait_for_transaction_receipt(txn)
print("Shipment registered!")
```

```python
# Store dummy IoT reading

txn = contract.functions.storeData(
    "RDG-TEST-001", # readingId
    "TEST-001", # rfidTag
    "GPS000", #deviceId
    "GPS", # deviceType
    "GPS", # dataType
    "14.5995,120.9842" # dataValue
).transact({'from': w3.eth.default_account, 'gas': 1000000})
w3.eth.wait_for_transaction_receipt(txn)
print("Dummy data stored on blockchain.")
```
